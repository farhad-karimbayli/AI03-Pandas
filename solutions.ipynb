## Задание 1: Анализ данных о книгах
**Условие:**
- У вас есть данные о книгах с полями: «Название», «Автор», «Год издания», «Цена».
- Создайте DataFrame с не менее чем шестью строками.
- Выведите весь DataFrame.
- Определите среднюю цену книг.
- Найдите книги, изданные после 2015 года.
- Отсортируйте DataFrame по цене в порядке возрастания.

```python
# 1. Создание DataFrame с данными о книгах (7 строк)
books_data = {
    "Название": [
        "Мастер и Маргарита",
        "1984",
        "Чистый код",
        "Грокаем алгоритмы",
        "Изучаем Python",
        "Преступление и наказание",
        "Глубокое обучение"
    ],
    "Автор": [
        "Михаил Булгаков",
        "Джордж Оруэлл",
        "Роберт Мартин",
        "Адитья Бхаргава",
        "Марк Лутц",
        "Фёдор Достоевский",
        "Ян Гудфеллоу"
    ],
    "Год издания": [1967, 1949, 2019, 2017, 2020, 1866, 2018],
    "Цена": [450.0, 380.0, 1250.0, 950.0, 1800.0, 350.0, 2100.0]
}

books_df = pd.DataFrame(books_data)
books_df
```

## Задание 2: Анализ данных о заказах из CSV
**Условие:**
- В CSV-файле содержатся данные о заказах: «Номер заказа», «Клиент», «Дата», «Сумма».
- Прочитайте данные из файла в DataFrame.
- Выведите первые десять строк.
- Определите количество заказов каждого клиента.
- Найдите максимальную и минимальную суммы заказов.
- Подсчитайте общую сумму всех заказов.

```python
# 1. Чтение данных из CSV-файла data/orders.csv
orders_df = pd.read_csv("data/orders.csv")

# 2. Вывод первых 10 строк
orders_df.head(10)
```

```python
# 3. Количество заказов каждого клиента
# Способ 1: value_counts() — самый компактный метод
print("Количество заказов (value_counts):")
print(orders_df["Клиент"].value_counts())

# Способ 2: groupby("Клиент").size()
# orders_df.groupby("Клиент")["Номер заказа"].count()
```

```python
# 4. Максимальная и минимальная суммы заказов
max_val = orders_df["Сумма"].max()
min_val = orders_df["Сумма"].min()
print(f"Максимальная сумма заказа: {max_val:.2f}")
print(f"Минимальная сумма заказа: {min_val:.2f}")

# Детали заказов с экстремальными суммами:
print("
Детали самого дорогого заказа:")
display(orders_df.loc[[orders_df["Сумма"].idxmax()]])
print("
Детали самого дешевого заказа:")
display(orders_df.loc[[orders_df["Сумма"].idxmin()]])
```

```python
# 5. Общая сумма всех заказов
total_sum = orders_df["Сумма"].sum()
print(f"Общая сумма всех заказов: {total_sum:.2f}")
```

## Задание 4: Сотрудники и проекты
**Условие:**
- Собраны данные о сотрудниках и их проектах: «Имя», «Проект», «Часы».
- Создайте DataFrame с не менее чем восьмью строками.
- Выведите исходный DataFrame.
- Подсчитайте общее количество часов по каждому сотруднику.
- Подсчитайте общее количество часов по каждому проекту.
- Найдите сотрудника, затратившего больше всего часов.

```python
# 1. Создание исходного DataFrame (9 строк)
employees_data = {
    "Имя": ["Али", "Али", "Лейла", "Лейла", "Мурад", "Мурад", "Айсель", "Эльвин", "Эльвин"],
    "Проект": [
        "Веб-платформа", "Мобильное приложение",
        "Веб-платформа", "Аналитика данных",
        "Мобильное приложение", "Информационная безопасность",
        "Аналитика данных",
        "Веб-платформа", "Информационная безопасность"
    ],
    "Часы": [40, 35, 30, 25, 45, 20, 50, 15, 30]
}

emp_df = pd.DataFrame(employees_data)
emp_df
```

```python
# 2. Общее количество часов по каждому сотруднику
hours_emp = emp_df.groupby("Имя")["Часы"].sum()
print("Часы по сотрудникам:")
print(hours_emp)
```

```python
# 3. Общее количество часов по каждому проекту
hours_proj = emp_df.groupby("Проект")["Часы"].sum()
print("Часы по проектам:")
print(hours_proj)
```

```python
# 4. Сотрудник, затративший больше всего часов
top_person = hours_emp.idxmax()
top_hours = hours_emp.max()
print(f"Больше всего часов затратил(а): {top_person} ({top_hours} ч.)")

# Вывод через сортировку:
hours_emp.sort_values(ascending=False).to_frame(name="Суммарные часы")
```

## Задание 5: Продажи билетов в кино
**Условие:**
- У вас есть таблица продаж билетов: «Фильм», «Город», «Проданные билеты».
- Создайте DataFrame с не менее чем двенадцатью строками.
- Выведите весь DataFrame.
- Подсчитайте общее количество проданных билетов по каждому фильму.
- Подсчитайте общее количество билетов по каждому городу.
- Найдите фильм с наибольшим количеством продаж.

```python
# 1. Создание DataFrame (12 строк)
tickets_data = {
    "Фильм": [
        "Интерстеллар", "Интерстеллар", "Интерстеллар",
        "Дюна 2", "Дюна 2", "Дюна 2",
        "Оппенгеймер", "Оппенгеймер", "Оппенгеймер",
        "Гладиатор 2", "Гладиатор 2", "Гладиатор 2"
    ],
    "Город": ["Баку", "Гянджа", "Сумгаит"] * 4,
    "Проданные билеты": [
        1200, 450, 600,
        2100, 850, 1100,
        1800, 720, 950,
        1500, 600, 800
    ]
}

tickets_df = pd.DataFrame(tickets_data)
tickets_df
```

```python
# 2. Общее количество проданных билетов по каждому фильму
sales_film = tickets_df.groupby("Фильм")["Проданные билеты"].sum()
print("Продажи по фильмам:")
print(sales_film)
```

```python
# 3. Общее количество билетов по каждому городу
sales_city = tickets_df.groupby("Город")["Проданные билеты"].sum()
print("Продажи по городам:")
print(sales_city)
```

```python
# 4. Фильм с наибольшим количеством продаж
top_film = sales_film.idxmax()
top_film_sales = sales_film.max()
print(f"Лидер проката: «{top_film}» с суммарными продажами {top_film_sales} билетов.")

# Рейтинг всех фильмов:
sales_film.sort_values(ascending=False).to_frame(name="Всего проданных билетов")
```
